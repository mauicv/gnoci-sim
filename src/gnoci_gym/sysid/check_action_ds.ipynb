```python
import json
import os

filename = os.getcwd() + '/dataset/dataset.json'

with open(filename, 'r') as f:
    data = json.load(f)
```

```python
import matplotlib.pyplot as plt
import numpy as np

actions_hw = np.array(data['data'][0]['actions_hw'])
actions_mjx = np.array(data['data'][0]['actions_mjx'])
plt.plot(actions_hw[:, 0])
```

```python
fig, axes = plt.subplots(2, 5, figsize=(12, 5))

for motor_idx, ax in enumerate(axes.flat):
    actions_mjx = np.array(data['data'][motor_idx]['actions_mjx'])[:, motor_idx]
    ax.plot(actions_mjx)
    ax.set_title(f"motor {motor_idx}")

plt.tight_layout()
plt.show()
```

```python
count = 0
for item in data['data']:
    if item['type'] == 'step':
        count += 1
        actions_hw = np.array(item['actions_hw'])
        plt.plot(actions_hw[:, item['motor_idx']])


plt.show()
assert count == 10 * 8, f"count is {count} but should be 10 * 8"
```

```python
for item in data['data']:
    if item['type'] == 'prbs':
        count += 1
        actions_hw = np.array(item['actions_hw'])
        actions_mjx = np.array(item['actions_mjx'])
        plt.plot(actions_mjx[:, item['motor_idx']])
        plt.plot(actions_hw[:, item['motor_idx']])
        break
```

```python
motor_idx_set = set()
count = 0
for item in data['data']:
    if item['type'] == 'prbs':
        count += 1
        motor_idx_set.add(item['motor_idx'])
        actions_hw = np.array(item['actions_hw'])
        plt.plot(actions_hw[:, item['motor_idx']])

assert motor_idx_set == set(range(10)), f"motor_idx_set is {motor_idx_set} but should be {set(range(10))}"

plt.show()
assert count == 10*3, f"count is {count} but should be 10 * 8"
```

```python
motor_idx_set = set()
count = 0
for item in data['data']:
    if item['type'] == 'ramp':
        count += 1
        motor_idx_set.add(item['motor_idx'])
        actions_hw = np.array(item['actions_hw'])
        plt.plot(actions_hw[:, item['motor_idx']])

assert motor_idx_set == set(range(10)), f"motor_idx_set is {motor_idx_set} but should be {set(range(10))}"

plt.show()
assert count == 10, f"count is {count} but should be 10 * 8"
```
